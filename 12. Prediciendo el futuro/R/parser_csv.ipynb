```python
import csv
import pandas as pd
```

```python
all_rows = []
with open('Index2020.csv', 'r') as csvfile:
    csv_reader = csv.reader(csvfile)
    next(csv_reader, None)
    for row in csv_reader:
        row_data = row[0].split(',')
        all_rows.append(row_data)
```

```python
for row in all_rows:
    row[1] = row[1].replace('"', '')
```

```python
df = pd.DataFrame(all_rows, columns=['id', 'date', 'spx', 'dax', 'ftse', 'nikkei'])

# Conversión de tipos
df = df.astype({
    'id': 'int32',
    'date': 'datetime64[ns]',
    'spx': 'float64',
    'dax': 'float64',
    'ftse': 'float64',
    'nikkei': 'float64'
})
```

```python
df.to_csv('Index2020_parser.csv', index=False)
```
