## Importing the relevant packages

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.graphics.tsaplots as sgt
import statsmodels.tsa.stattools as sts
from statsmodels.tsa.arima.model import ARIMA
from scipy.stats.distributions import chi2 
from math import sqrt
import seaborn as sns
sns.set()
```

## Importing the Data and Pre-processing 

```python
raw_csv_data = pd.read_csv("Index2018.csv") 
df_comp=raw_csv_data.copy()
df_comp.date = pd.to_datetime(df_comp.date, dayfirst = True)
df_comp.set_index("date", inplace=True)
df_comp=df_comp.asfreq('b')
df_comp=df_comp.ffill()
```

```python
df_comp['market_value']=df_comp.ftse
```

```python
del df_comp['spx']
del df_comp['dax']
del df_comp['ftse']
del df_comp['nikkei']
size = int(len(df_comp)*0.8)
df, df_test = df_comp.iloc[:size], df_comp.iloc[size:]
```

```python
import warnings
warnings.filterwarnings("ignore")
```

## The LLR Test

```python
def LLR_test(mod_1, mod_2, DF = 1):
    L1 = mod_1.llf
    L2 = mod_2.llf
    LR = (2*(L2-L1))    
    p = chi2.sf(LR, DF).round(3)
    return p
```

## Creating Returns

```python
df['returns'] = df.market_value.pct_change(1)*100 # porcentaje de cambio, mide que tanto cambia el precio siguiente respecto el anterior.
```

```python
df.head()
```

## Creating Squared Returns

```python
df['sq_returns'] = df.returns.mul(df.returns)
```

```python
df.sq_returns.plot(figsize=(20,5))
plt.title("Volatility", size = 24)
plt.show()
```

```python
sgt.plot_pacf(df.returns[1:], lags = 40, alpha = 0.05, zero = False , method = ('ols'))
plt.title("PACF of Returns", size = 20)
plt.show()
```

No se parece a ruido blanco

```python
sgt.plot_pacf(df.sq_returns[1:], lags = 40, alpha = 0.05, zero = False , method = ('ols'))
plt.title("PACF of Squared Returns", size = 20)
plt.show()
```

En general la regla dicta que si la seríe temporal se asemeja a ruido blanco, mientras que su versión cuadrada sugiere usar un modelo AR(n), debemos recurrir a un modelo ARCH.(n) para modelar la varianza.

## The Simple ARCH(1) 

```python
model_arch_1 = arch_model(df.returns[1:], mean = "Constant", vol = "ARCH", p = 1) # p es el número de rezagos de la varianza.
# mean = "Constant" indica que el modelo tiene una media constante.
# vol = "ARCH" se establece el modelo para la volatilidad
# p = 1 indica que el modelo tiene una varianza que depende de los rezagos de la varianza de orden 1.
results_arch_1 = model_arch_1.fit(update_freq = 1)
results_arch_1.summary()
```

Tenemos un modelo ARCH(1) con medía constante, R-squared y Adj. R-squared son cero, estas medidas de la variación explicadas por el modelo, a diferencia de los modelos AR como tenemos una media constante, no hay nada que explicar respecto a la variabilidad de los residios, en estos modelos estas medidas no son utiles.
Log-Likelihood:	-7660.09. Mide la bondad de ajuste del modelo y es mas significativa que los modelos estudiados anteriormenete para nuestros datos.
Las demas observaciones se analizan de forma analoga a estudios previos.

**Tener en cuenta que estos modelos solo se utiliza para predicir variaciones futuras de la serie en lugar de rendimientos futuros, si nos intereza medir la estabilidad de los rendimientos lo podemos usar perfectamente, no si nos interesa medir la dirección del precio**

## Higher-Lag ARCH Models

```python
model_arch_2 = arch_model(df.returns[1:], mean = "Constant", vol = "ARCH", p = 2)
results_arch_2 = model_arch_2.fit(update_freq = 5)
results_arch_2.summary()
```

ARCH(2) supera al ARCH(1)
